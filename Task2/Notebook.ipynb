## 🟢 Data Preprocessing

```python
from sklearn.preprocessing import StandardScaler , RobustScaler
```

```python
sqft_living = df["sqft_above"] + df["sqft_basement"]
true_Values = (sqft_living == df["sqft_living"]).value_counts().values

print("'sqft_living' = 'sqft_above' + 'sqft_basement' " if true_Values == df.shape[0] else "")
```

```python
df.drop(columns=["sqft_above","sqft_basement"],inplace = True)
```

### 🟢 Update `Bathrooms` Feature to int numbers 

```python
df["bathrooms"] = df["bathrooms"].apply(lambda x:np.ceil(x))
```

### 🟢 Make New Feature from `yr_renovated` & `yr_built`

```python
df['house_age'] = df['sale_year'] - df['yr_built']
```

```python
df['yr_modified'] = np.where(df['yr_renovated'] == 0, df['yr_built'], df['yr_renovated'])
df['years_since_modification'] = df['sale_year'] - df['yr_modified']

df.drop(columns=['yr_built', 'yr_renovated', 'yr_modified'], inplace=True)
```

### 🟢 Scaling Features

```python
scalers = {}
normalized = ["sqft_living15","sqft_lot15","sale_year","house_age","years_since_modification"] 

for column in normalized:
    scaler = RobustScaler()
    df[column] = scaler.fit_transform(df[[column]])
    scalers[column] = scaler
```

```python
df["price"] = np.log1p(df["price"])
df['sqft_living'] = np.log1p(df['sqft_living'])
df['sqft_lot'] = np.log1p(df['sqft_lot'])
```

### 🟢 Encoding `Zipcode` 

```python
zipcode_target_mean = df.groupby('zipcode')['price'].median()

df['zipcode_encoded'] = df['zipcode'].map(zipcode_target_mean)

df.drop(columns=['zipcode'], inplace=True)
```

## 🛠 Building Model 

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error , mean_absolute_error , root_mean_squared_error , r2_score
```

```python
def Print_Scores(Scores:dict):
    for i,(Score_type , score) in enumerate(Scores.items(),start=1):
        print(f"{i}. {Score_type} = {score:0.4f}" if i != 1 else f"{i}. {Score_type} = {score*100:0.2f} %")
```

```python
def Print_Features_Weights(Weights:list,Features:list):
    for i, (weight , feature) in enumerate(zip(Weights,Features),start=1):
        print(f"{i}. '{feature}' Weight = {weight:0.2f}")
```

```python
X = df.drop(columns=["price"])
Y = df["price"]
```

```python
x_train,x_test,y_train,y_test = train_test_split(X,Y,test_size=0.2,shuffle=True,random_state=42)
```

## ✅ Linear Regression

```python
model = LinearRegression()
model.fit(x_train,y_train)
y_predicted = model.predict(x_test)
```

```python
Scores = {
    "R2 Score":r2_score(y_true=y_test,y_pred=y_predicted),
    "MSE":mean_squared_error(y_true=y_test,y_pred=y_predicted),
    "MAE":mean_absolute_error(y_true=y_test,y_pred=y_predicted),
    "RMSE":root_mean_squared_error(y_true=y_test,y_pred=y_predicted)
}
Print_Scores(Scores=Scores)
```

```python
Print_Features_Weights(Weights=model.coef_.tolist(),Features=x_train.columns.tolist())
```

## ✅ Random Forest 

```python
model = RandomForestRegressor(n_estimators=100,max_samples=0.5)
model.fit(x_train,y_train)
y_predicted = model.predict(x_test)
```

```python
Scores = {
    "R2 Score":r2_score(y_true=y_test,y_pred=y_predicted),
    "MSE":mean_squared_error(y_true=y_test,y_pred=y_predicted),
    "MAE":mean_absolute_error(y_true=y_test,y_pred=y_predicted),
    "RMSE":root_mean_squared_error(y_true=y_test,y_pred=y_predicted)
}
Print_Scores(Scores=Scores)
```

```python
Print_Features_Weights(Weights=model.feature_importances_.tolist(),Features=x_train.columns.tolist())
```
